## 6) Limpieza de datos

```r
# ============================
# CARGA DE LIBRERÍAS
# dplyr facilita manipulación de datasets
# janitor ayuda a limpiar nombres de variables
# ============================

if (!require(dplyr, quietly = TRUE)) {
  suppressMessages(install.packages("dplyr"))
}
library(dplyr)

if (!require(janitor, quietly = TRUE)) {
  suppressMessages(install.packages("janitor"))
}
library(janitor)
```

```r
library(tidyverse)
library(stringr)

# Estandarización de texto, limpieza de errores y eliminación de NAs
dataset_limpio <- data_filtrado %>%
  # 0. Renombrar PRIMERO todas las columnas al español
  rename(
    id_paciente = Patient.ID,
    frecuencia_cardiaca = Heart.Rate,
    frecuencia_respiratoria = Respiratory.Rate,
    fecha_hora = Timestamp,
    temperatura_corporal = Body.Temperature,
    saturacion_oxigeno = Oxygen.Saturation,
    presion_arterial_sistolica = Systolic.Blood.Pressure,
    presion_arterial_diastolica = Diastolic.Blood.Pressure,
    edad = Age,
    genero = Gender,
    peso_kg = Weight..kg.,
    altura_m = Height..m.,
    vfc_derivada = Derived_HRV,
    presion_pulso_derivada = Derived_Pulse_Pressure,
    imc_derivado = Derived_BMI,
    pam_derivada = Derived_MAP,
    categoria_riesgo = Risk.Category,
    diagnostico = Diagnosis,
    estado_fumador = Smoking.Status,
    nivel_actividad_fisica = Physical.Activity.Level,
    tipo_medicacion = Medication.Type,
    region = Region,
    nivel_socioeconomico = Socioeconomic.Level,
    ocupacion = Occupation,
    turno = Shift,
    estacion = Season,
    glucosa = Glucose,
    colesterol = Cholesterol,
    trigliceridos = Triglycerides,
    hemoglobina = Hemoglobin,
    creatinina = Creatinine
  ) %>%

  # 1. Pasar todo el texto a minúsculas y quitar espacios extra a los lados
  mutate(across(where(is.character), ~ str_to_lower(str_trim(.)))) %>%

  # 1.5. Transformar celdas con "error_" en NA reales en TODAS las variables (incluye categoria_riesgo)
  mutate(across(where(is.character), ~ ifelse(str_detect(., "^error_"), NA, .))) %>%

  # 2. Corregir y unificar categorías específicas
  mutate(
    genero = case_when(
      genero %in% c("female", "femenino", "f") ~ "femenino",
      genero %in% c("male", "masculino", "m")   ~ "masculino",
      TRUE ~ genero
    ),

    nivel_actividad_fisica = case_when(
      nivel_actividad_fisica %in% c("high", "h1gh")        ~ "alto",
      nivel_actividad_fisica %in% c("moderate", "modrate", "") ~ "medio",
      nivel_actividad_fisica %in% c("low", "lwo")          ~ "bajo",
      TRUE ~ nivel_actividad_fisica
    ),

    tipo_medicacion = case_when(
      tipo_medicacion %in% c("beta blockers", "beta-blockers")           ~ "bloqueadores_beta",
      tipo_medicacion %in% c("anti hypertensive", "antihypertensives",
                            "antihypertensive")                          ~ "antihipertensivos",
      tipo_medicacion %in% c("insulin", "insuline")                      ~ "insulina",
      tipo_medicacion %in% c("none") ~ "ninguna",
      TRUE ~ tipo_medicacion
    ),

    region = case_when(
      region %in% c("urban")                      ~ "urbana",
      region %in% c("rural")                      ~ "rural",
      region %in% c("semiurban", "semi-urban", "semi urban")                 ~ "semiurbana",
      TRUE ~ region
    ),

    ocupacion = case_when(
      ocupacion %in% c("student")                ~ "estudiante",
      ocupacion %in% c("office worker", "office") ~ "oficinista",
      ocupacion %in% c("retired", "retird")      ~ "jubilado",
      ocupacion %in% c("manual labor")          ~ "trabajo_manual",
      TRUE ~ ocupacion
    ),

    diagnostico = case_when(
      diagnostico == "healthy"         ~ "sano",
      diagnostico == "hypertension"    ~ "hipertensión",
      diagnostico == "diabetes"        ~ "diabetes",
      diagnostico == "cardiac disease" ~ "enfermedad_cardiaca",
      diagnostico == "copd"            ~ "epoc",
      TRUE ~ diagnostico
    ),

    estado_fumador = case_when(
      estado_fumador == "never"   ~ "nunca",
      estado_fumador == "former"  ~ "ex_fumador",
      estado_fumador == "current" ~ "fumador_actual",
      TRUE ~ estado_fumador
    ),

    nivel_socioeconomico = case_when(
      nivel_socioeconomico == "low"    ~ "bajo",
      nivel_socioeconomico == "middle" ~ "medio",
      nivel_socioeconomico == "high"   ~ "alto",
      TRUE ~ nivel_socioeconomico
    ),

    categoria_riesgo = case_when(
      categoria_riesgo == "high risk" ~ "alto_riesgo",
      categoria_riesgo == "low risk"  ~ "bajo_riesgo",
      TRUE ~ categoria_riesgo
    ),

    turno = case_when(
      turno == "morning"   ~ "mañana",
      turno == "afternoon" ~ "tarde",
      turno == "night"     ~ "noche",
      TRUE ~ turno
    ),

    estacion = case_when(
      estacion == "winter" ~ "invierno",
      estacion == "spring" ~ "primavera",
      estacion == "summer" ~ "verano",
      estacion == "autumn" ~ "otoño",
      TRUE ~ estacion
    )
  ) %>%

  # 3. ACTUALIZACIÓN: Recalcular la cantidad de NAs por fila usando cur_data()
  mutate(cantidad_na = rowSums(is.na(cur_data()))) %>%
  filter(cantidad_na < 7) %>%

  # 4. Eliminar duplicados
  distinct()

# Revisar cuántas filas nos quedaron y un resumen rápido de NAs
cat("Total de registros válidos:", nrow(dataset_limpio), "\n")
table(dataset_limpio$cantidad_na)
```

### PREPARACIÓN DE VARIABLES NUMÉRICAS

```r
# ============================
# PREPARACIÓN DE VARIABLES NUMÉRICAS
# Convertir variables a numéricas y formatear resultados
# ============================
```

```r
summary(dataset_limpio)
```

```r

# ============================
# CONVERTIR CHARACTER A NUMÉRICO
# as.numeric() transforma texto en números
# Útil cuando R leyó números como character
# ============================

class(dataset_limpio$glucosa)

dataset_limpio <- dataset_limpio %>%
  mutate(
    glucosa = as.numeric(glucosa)
  )

class(dataset_limpio$glucosa)

# ============================
# CONVERTIR VARIAS VARIABLES A NUMÉRICAS
# across() aplica la misma transformación
# ============================

dataset_limpio <- dataset_limpio %>%
  mutate(
    across(
      c(glucosa, colesterol, trigliceridos, hemoglobina, creatinina),
      as.numeric
    )
  )
```

```r
# ============================
# DESACTIVAR NOTACIÓN CIENTÍFICA
# scipen alto fuerza formato decimal
# ============================

options(scipen = 999)
```

```r
# ============================
# REDONDEAR A 2 DECIMALES TODAS LAS VARIABLES NUMÉRICAS
# round(x, 2)
# ============================

dataset_limpio <- dataset_limpio %>%
  mutate(
    across(
      where(is.numeric),
      ~ round(., 2)
    )
  )
```

```r
summary(dataset_limpio)
```

```r
# ============================
# VALIDACIÓN CLÍNICA: HEART RATE
# Frecuencia cardíaca (bpm)
# Rango clínicamente plausible: 30 - 220
# ============================
dataset_limpio %>%
  summarise(
    minimo = min(frecuencia_cardiaca, na.rm = TRUE),
    maximo = max(frecuencia_cardiaca, na.rm = TRUE)
  )

  # ============================
# IDENTIFICAR OUTLIERS CLÍNICOS
# Valores fuera del rango aceptable
# ============================

outliers_hr <- dataset_limpio %>%
  filter(frecuencia_cardiaca < 30 | frecuencia_cardiaca > 220)

outliers_hr

# ============================
# ELIMINAR OUTLIERS DE HEART RATE
# Conserva solo valores dentro del rango clínico
# ============================

dataset_limpio <- dataset_limpio %>%
  filter(frecuencia_cardiaca >= 30,
         frecuencia_cardiaca <= 220)
```

```r
# Cargamos lubridate para manejar fechas fácilmente
library(lubridate)

# Ajuste de tipos de datos sobre nuestro dataset_limpio
dataset_limpio <- dataset_limpio %>%
  # A. Convertir la columna de texto a formato fecha-hora real
  mutate(
    fecha_hora = ymd_hms(fecha_hora)
  ) %>%

  # B. Convertir variables cualitativas a "Factores" (categorías)
  mutate(
    across(
      c(genero, categoria_riesgo, diagnostico, estado_fumador,
        nivel_actividad_fisica, tipo_medicacion, region,
        nivel_socioeconomico, ocupacion, turno, estacion),
      as.factor
    )
  )

# Verificamos cómo cambió la estructura
str(dataset_limpio %>% select(fecha_hora, genero, region))
```

```r
# ==========================================
# PARTE 6 - INGENIERÍA DE VARIABLES
# ==========================================
# ==========================================
# VARIABLE DERIVADA 1: Indicador de Anemia
# ==========================================

dataset_limpio <- dataset_limpio %>%
  mutate(
    indicador_anemia = case_when(
      genero == "femenino" & hemoglobina < 12 ~ 1,
      genero == "masculino" & hemoglobina < 13 ~ 1,
      TRUE ~ 0
    )
  )
# ==========================================
# VARIABLE DERIVADA 2: Indicador de Taquicardia
# ==========================================

dataset_limpio <- dataset_limpio %>%
  mutate(
    indicador_taquicardia = as.numeric(frecuencia_cardiaca >= 100)
  )
  # ==========================================
# VARIABLE DERIVADA 3: Colesterol Alto
# ==========================================

dataset_limpio <- dataset_limpio %>%
  mutate(
    colesterol_alto = as.numeric(colesterol >= 240)
  )
  dataset_limpio %>% select(indicador_anemia, indicador_taquicardia, colesterol_alto) %>% head(10)
```

```r
# ==========================================
# RECOLECCIÓN DE RESULTADOS PARA LA GUÍA
# ==========================================

# 1. Dimensiones finales
cat("=== DIMENSIONES FINALES ===\n")
cat("Filas:", nrow(dataset_limpio), "\n")
cat("Columnas:", ncol(dataset_limpio), "\n\n")

# 2. Pacientes únicos
cat("=== PACIENTES ÚNICOS ===\n")
cat("Pacientes únicos:", n_distinct(dataset_limpio$id_paciente), "\n\n")

# 3. NAs por columna (solo las que tienen)
cat("=== NAs POR COLUMNA ===\n")
na_count <- colSums(is.na(dataset_limpio))
print(na_count[na_count > 0])
cat("\n")

# 4. Rangos de variables numéricas (para validación clínica)
cat("=== RANGOS DE VARIABLES NUMÉRICAS ===\n")
numericas <- dataset_limpio %>% select(where(is.numeric))
rangos <- data.frame(
  Variable = names(numericas),
  Min = sapply(numericas, min, na.rm = TRUE),
  Max = sapply(numericas, max, na.rm = TRUE)
)
print(rangos)
cat("\n")

# 5. Distribución de variables categóricas clave
cat("=== CATEGORÍA DE RIESGO ===\n")
print(table(dataset_limpio$categoria_riesgo, useNA = "always"))
cat("\n")

cat("=== DIAGNÓSTICO ===\n")
print(table(dataset_limpio$diagnostico, useNA = "always"))
cat("\n")

cat("=== GÉNERO ===\n")
print(table(dataset_limpio$genero, useNA = "always"))
cat("\n")

cat("=== ESTADO FUMADOR ===\n")
print(table(dataset_limpio$estado_fumador, useNA = "always"))
cat("\n")

cat("=== NIVEL DE ACTIVIDAD FÍSICA ===\n")
print(table(dataset_limpio$nivel_actividad_fisica, useNA = "always"))
```

```r
# ==========================================
# VARIABLE DERIVADA 4: Categoria de creatinina
# Clasifica la funcion renal segun la creatinina serica.
# Normal < 1,00 | Leve 1,00-1,49 | Moderada 1,50-2,99 | Severa >= 3,00
# ==========================================

dataset_limpio <- dataset_limpio %>%
  mutate(
    categoria_creatinina = cut(
      creatinina,
      breaks = c(-Inf, 0.999, 1.499, 2.999, Inf),
      labels = c("Normal", "Leve", "Moderada", "Severa")
    )
  )

table(dataset_limpio$categoria_creatinina)
```

## Validación final y exportación

Cierre de la limpieza: se descarta el único registro que no puede usarse en el
análisis y se exporta el dataset definitivo, de **13.331 registros**, que es el
que consume el archivo de análisis.


```r

cat("Registros antes de la validacion final:", nrow(dataset_limpio), "\n")

registros_invalidos <- dataset_limpio %>%
  filter(is.na(id_paciente) | is.na(genero) | is.na(categoria_riesgo))

cat("Registros sin identificacion o sin etiqueta clinica:", nrow(registros_invalidos), "\n")
registros_invalidos

dataset_limpio <- dataset_limpio %>%
  filter(!is.na(id_paciente), !is.na(genero), !is.na(categoria_riesgo))

cat("DATASET FINAL:", nrow(dataset_limpio), "registros y",
    ncol(dataset_limpio), "variables\n")
cat("Pacientes unicos:", length(unique(dataset_limpio$id_paciente)), "\n")
```
